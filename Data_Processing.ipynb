# Data preprocessing

This notebook contains preprocessing of the four data tables we have chosen. To check the data quality and missing values, we used Sweetviz. There were no missing values:) We decided to split each dataset in a 7:3 ratio (train:test) and later scale all values in each new dataset using the scale_dataframes function.

```python
import numpy as np
import pandas as pd

# import sweetviz as sv - only to check data quality
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.model_selection import train_test_split
```

```python
def scale_dataframes(train_df, test_df, target_column):
  
    numeric_cols = [col for col in train_df.columns if col != target_column]
    
    scaler = StandardScaler()
    scaler.fit(train_df[numeric_cols])
    
    train_scaled = train_df.copy()
    test_scaled = test_df.copy()
    
    train_scaled[numeric_cols] = scaler.transform(train_df[numeric_cols])
    test_scaled[numeric_cols] = scaler.transform(test_df[numeric_cols])
    
    return train_scaled, test_scaled
```

## Heart Attack dataframe

 Target variable -> class

```python
heart_df = pd.read_csv('Data/Heart-Attack.csv')
heart_df.head()
```

```python
# raport_1 = sv.analyze(heart_df)
# raport_1.show_html("raport_1_sweetviz.html")
```

```python
heart_df['class'] = heart_df['class'].map({'negative': 0, 'positive': 1})
```

```python
heart_train_df, heart_test_df = train_test_split(heart_df, test_size=0.25, stratify=heart_df['class'], random_state=42)
```

```python
heart_train_df, heart_test_df = scale_dataframes(heart_train_df, heart_test_df, target_column='class')
```

```python
heart_train_df.to_csv('Data/heart_train.csv', index=False)
heart_test_df.to_csv('Data/heart_test.csv', index=False)
```

## Mushrooms dataframe

Target Variable -> class

All values are categorical, so we use OneHotEncoder and we don't scale it.

```python
mushrooms_df = pd.read_csv('Data/mushrooms.csv')
mushrooms_df.head()
```

```python
# raport_2 = sv.analyze(mushrooms_df)
# raport_2.show_html("raport_2_sweetviz.html")
```

```python
mushrooms_df['class'] = mushrooms_df['class'].map({'p': 0, 'e': 1})
```

```python
y = mushrooms_df["class"]
X = mushrooms_df.drop(columns=['class'])

encoder = OneHotEncoder(sparse_output=False)  
X_encoded = encoder.fit_transform(X)

X_encoded_df = pd.DataFrame(X_encoded, columns=encoder.get_feature_names_out(X.columns))
mushrooms_df = pd.concat([X_encoded_df, y.reset_index(drop=True)], axis=1)
```

```python
mushrooms_train_df, mushrooms_test_df = train_test_split(mushrooms_df, test_size=0.25, stratify=mushrooms_df['class'], random_state=42)
```

```python
mushrooms_train_df, mushrooms_test_df = scale_dataframes(mushrooms_train_df, mushrooms_test_df, target_column='class')
```

```python
mushrooms_train_df.to_csv('Data/mushrooms_train.csv', index=False)
mushrooms_test_df.to_csv('Data/mushrooms_test.csv', index=False)
```
